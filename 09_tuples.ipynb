```python
# Tuples
# Tuples are ordered collection of items that are immutable. They are similar 
# to lists, but their immutability makes them different.
# Tuples are used to store multiple items in a single variable.
# A tuple is a collection which is ordered and unchageable.
# Tuples are unchangeable and allow duplicate values
```

```python
# creating a tuple
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))

fruits_data = ("apple", "bananaa")
print(len(fruits_data))
```

```python
# check if item exists

laptop_data = ("lenovo", "dell", "axus")

if "lenovo" in laptop_data:
    print("lenovo is there in laptop_data")
```

```python
# Loop Through a Tuple

laptop_data2 = ("lenovo", "dell", "axus")

for x in laptop_data2:
    print(x)

```

```python
# joining two tuples

tup1 = ("lenovo", "dell", "axus")
tup2 = (1, 2, 3)

output = tup1 + tup2
print(output)
```

```python
# tuple constructor

fru = tuple(("apple", "banana", "pineapple"))
print(fru)
print(fru[-1])
```

```python
num = tuple([1, 2, 3, 4, 5])
print(num)
```

```python
num2 = list((1, 2, 3, 4, 5))
print(num2)
print(type(num2))
```

```python
mixed_tuple = (1, True, "pradeep kumar kairamkonda", 1.15)
print(mixed_tuple)

# Accessing tuple elements

result = mixed_tuple[0]
print(result)
print(mixed_tuple[2])
print(mixed_tuple[-1])
print(mixed_tuple[0 : 3])
print(mixed_tuple[::-1])  # reverse the tuple
print(mixed_tuple[::2])
```

```python
# tuple operations

data1 = (1, 2, 3, 4, 5)
data2 = (6, 7, 8, 9, 10)

result = data1 + data2
print(result)

result2 = result * 3
print(result2)
```

```python
# Immutable nature of tuples
# Tuples are immutable, meaning their elements cannot be changed once assigned

list_data = [1, 2, 3, 4, 5]
print(list_data)

list_data[0] = "pradeep kumar kairamkonda"
print(list_data)     # list are mutable

student_id = (1, 2, 3, 4, 5)
print(student_id)

# tuples are immutable. the elements cannot be change once assigned

# student_id[2] = "pradeep kumar kairamkonda"
# print(student_id)
```

```python
# Tuple mehtods

student_data = (1, 2, 3, 4, 5, "pradeep kumar kairamkonda")
print(student_data.count(1))
print(student_data.index("pradeep kumar kairamkonda"))
 
```

```python
# packing and unpacking tuple
# packing
packed_tuple = 1, "pradeep kumar kairamkonda", True, 3.14
print(packed_tuple)

# unpacking a tuple

fruits = ("orange", "grapes", "banana")

a, b, c = fruits
print(a)
print(b)
print(c)

# unpacking with *

friends = ("mani rathan", "prashanth", "sam", "arun")

x, *y, z = friends
print(x)
print(y)
print(z)
```

```python
# nested tuple
# tuples inside a tuple is called a nested tuples

lst = [[1, 2, 3, 4], [5, 6, 7, 8], [1, 3.14, "pradeep", True]]
print(lst[0])
print(lst[0][2])
print(lst[2][0 : 3])

# inside the list also we can create the tuples

lst2 = [[1, 2, 3, 4], [5, 6, 7, 8], (1, 3.14, "pradeep", True)]
print(lst2)
print(lst2[2])
```

```python
# nested tuples
data4 = (("kumar", "netha", "reddy"), (1, 2, 3), (True, False))
print(data4)
print(data4[0])
print(data4[0][2])
```
