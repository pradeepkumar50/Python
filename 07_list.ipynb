```python
# List

# List are ordered, mutable collections of items
# They can contain items of different data types

# Lists are used to store multiple items in a single variable

# There are 4 built in data types in python

# List
# Tuple
# set
# Dictionary
```

```python
fruits = ["apple", "pineapple", "canberry", "rasberry","grapes"]
print(fruits)
print(fruits[0])

# In list item starts with index 0
# list is a muteable

fruits[3] = "pradeep kumar kairamkonda"
print(fruits)

# list length

print(len(fruits))

list1 = ["pradeep", "kumar", 25, True]
print(list1)
print(type(list1))

# Finding a last element in the list

print(list1[len(list1) - 1])

# The list() constructor

list2 = list(("apple","blueberry","canberry"))
print(list2)

# Accessing a list

name = ["pradeep", "kumar", "kairamkonda"]
print(name[0])

# negitive indexing
# It means strats from end

print(name[-1])

# Range of indexes
data = [1, 2, 3, 4, "pradeep", "kumar", "kairamkonda", 5, 6, 7, 8, 9, 10]
print(data[4:7])

# The search will start at index 4 and end at inex 6

# The items returns from the beginning

print(data[:4])

# The items returns from index 3  to the end

print(data[3:])
print(data[::2])  # step
print(data[::-1])
```

```python
# Examples for negitve -4 to but not including -1

fruits = ["apple", "banana", "canberry","blueberry", "rasberry", "strawberry"]

print(fruits[-4 :-1])

# check if item exists

if "banana" in fruits:
    print("yes, banana is there in the fruits list")
```

```python
# changing the item value

fruits_list = ["apple", "banana", "canberry","blueberry", "rasberry"]

fruits_list[1] = "pineapple"
print(fruits_list)

# changing the range of item values

fruits_list[2 : 4] = ["watermelon", "mango"]
print(fruits_list)

# changing the second value by replacing with the two new values

fruits_list[0 : 1] = ["pradeep", "kumar"]
print(fruits_list)
```

```python
# insert items
# the insert() method inserts an item at the specified index

num = [1, 2, 3, 4, 5]
num.insert(2, "pradeep kumar kairamkonda")
print(num)
```

```python
# Append
# To add the item at the end of the list

data = [1, 2, 3, 4, "pink", 5, 2, "pink"]
data.append("pradeep")
print(data)

# insert item inserts at the specified index

data.insert(2, "kumar")
print(data)

# the remove() method removes the specified item

data.remove("pink")
print(data)

# It removes the first occurance of pink

```

```python
# pop method removes the last element in the list

name = ["pradeep", "kumar", "kairamkonda"]
name.pop()
print(name)
```

```python
# del also removes the element

num = ["one", "two", "three"]
del num[0]
print(num)
```

```python
# del also removes the whole list
data = ["one", "two", "three"]
del data
# print(data)

# rsult causes error
```

```python
# index method

devices = ["lenovo", "ipad", "tv"]
result = devices.index("ipad")
print(result)
```

```python
# sort the list in ascending order
fruits_list2 = ["apple", "banana", "canberry","blueberry", "rasberry","avacodo"]
fruits_list2.sort()
print(fruits_list2)

# reverse method reverse the list
fruits_list2.reverse()
print(fruits_list2)

# clear() clears the list
fruits_list2.clear()
print(fruits_list2)

```

```python
# Iterating the list

fruits_list3 = ["apple", "banana", "canberry","blueberry", "rasberry","avacodo"]

for item in fruits_list3:
    print(item)

# Iterating with index

for index, item in enumerate(fruits_list3 ):
    print(item, index)
```

```python
# List comprehension

list_data = []
for x in range(10):
    list_data.append(x ** 2)

print(list_data)
```

```python
# list compreshnion syntax
# newlist = [expression for item in iterable if condition == True]

data = [x ** 2 for x in range(10)]
print(data)

# list comprehension

# Basic syntax             [expression for item in itrable ]
# with conditionallogic    [ expression for item in iterable if cndition]
```

```python
# Basic list compresion

square = [numbers ** 2 for numbers in range(20)]
print(square)
```

```python
even_data = []
for i in range(10):
    if i % 2 == 0:
        even_data.append(i)
print(even_data)
```

```python
num_data = [num for num in range(20) if num %2 == 0]
print(num_data)
```
