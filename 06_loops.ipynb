```python
# loops
# Python has two primitive loop commands

# for loop
# while loop

# loop control statements

# break
# continue
# pass

# Nested loops
```

```python
# for loop

# A for loop is used for iterating over a sequence i.e either a list, a tuple,
# a dictionary, a set, or a string.

# With the for loop we can execute a set of statements, once for each item in
# list, tuple, set etc.

for i in range(5):
    print(i)

```

```python
for i in range(1,10):
    print(i)
```

```python
# start, stop, step parameter 
for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(10, 1, -1):
    print(i)
```

```python
for i in range(10, 1, -2):
    print(i)
```

```python
name = ["pradeep", "kumar", "kairamkonda"]
for i in name:
    print(i)
```

```python
# strings

name = "pradeep kumar kairamkonda"
for i in name:
    print(i)
```

```python
# while loop

# The while loop continues to execute as long as the condition is true

count = 0

while count < 5:
    print(count)
    count = count + 1
    
```

```python
count = 0

while count % 2 == 0:
    print(count)
    count = count + 1
```

```python
# loop control statements

# break
# In break statement we can stop the loop even if the condiiton is true

for i in range(10):
    if(i == 6):
        break
    print(i)
```

```python
# break in while loop

i = 2

while i < 10:
    print(i)
    if i == 5:
        break
    i = i + 1
```

```python
# continue

# The continue statement skips the current iteration and continues with the next

for i in range(10):
    if i == 4:
        continue
    print(i)

```

```python
i = 2
while i < 10:
    i = i + 1
    if i == 5:
        continue
    print(i)
```

```python
# pass
# The pass statement is a null operation; It does nothing

for i in range(5):
    if i == 3:
        pass
    print(i)



```

```python
# nested loops

# A loop inside a loop is called nested loop

my_name = ["pradeep", "kumar", "kairamkonda"]
friend_name = ["pramod", "reddy", "bommidi"]

for i in my_name:
    for j in friend_name:
        print(i, j)

```
