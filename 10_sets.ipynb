```python
# sets

# sets are a built in data type in python used to store collections of unique
# items. they are unordered means the elements do not follow the specific order
# they do no allow the dublicate elements

# A set is a collection which is unordered, unchangable and unindexed
# set items are unchangle, but you can remove items and add new items
```

```python
# creating a set

fruits = {"apple", "banana", "cherry"}
print(fruits)

for x in fruits:
    print(x)

fruits.add("watermelon")
print(fruits)
```

```python
# dublicates are not allowed

laptops = {"hp", "dell", "lenovo", "hp"}
print(laptops)
print(type(laptops))
```

```python
num_set = set([1, 2, 3, 4, 5])
print(num_set)

num_set2 = set((1, 2, 3, 4, 5))
print(num_set2)
```

```python
# dublicates elements

num_set2 = set((1, 2, 3, 4, 5, 3, 2))
print(num_set2)

# Basics sets operations
# Adding and removing the elements

num_set2.add(10)
print(num_set2)

# adding the same element again it will ignore
num_set2.add(10)
print(num_set2)

# removing the elements

num_set2.remove(5)
print(num_set2)

# discard method : remove an element from a set if it is a member

num_set2.discard(20)
print(num_set2)

# pop method
# first in first out  ------- removes first element
num_set2.pop()
print(num_set2)

# clear all the elements

num_set2.clear()
print(num_set2)
```

```python
# Set Membership test
# If the element is present it returns true otherwise it returns false

num_set3 = {1, 2, 3, 4, 5}
print(2 in num_set3)
print(10 in num_set3)
```

```python
# Mathematical operations

num_set4 = {1, 2, 3, 4, 5, 6}
num_set5 = {4, 5, 6, 7, 8, 9}

# Union 

union_set = num_set4.union(num_set5)
print(union_set)

# Intersection ---- common elements from both the sets

intersection_set = num_set4.intersection(num_set5)
print(intersection_set)

# Intersection update ---- common elements update in num_set4

num_set4.intersection_update(num_set5)
print(num_set4)

```

```python
# Difference
# It removes common elements in both the sets

num_set6 = {1, 2, 3, 4, 5, 6}
num_set7 = {4, 5, 6, 7, 8, 9}

difference_set = num_set6.difference(num_set7)
print(difference_set)

# symmetric difference

sym_set = num_set6.symmetric_difference(num_set7)
print(sym_set)   # both the unique elements form the both sets are combined
```

```python
# sets methods

num_set8 = {1, 2, 3}
num_set9 = {3, 4, 5}

# is subset    ---- All the elements avialble in set

res = num_set8.issubset(num_set9)
print(res)

res2 = num_set8.issuperset(num_set9)
print(res2)
```

```python
lst = [1, 2, 3, 4, 5, 2, 3, 4]
print(set(lst))
```

```python
# frozenset

# frozenset() constructor to create a frozenset from any iterable.

fruits_data = frozenset({"peach", "watermelon", "orange"})
print(fruits_data)
print(type(fruits_data))
```
