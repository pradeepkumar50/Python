```python
# list comprehnsion
# comprehension is a process of creating new sequence from existing sequence

even_data = []
for i in range(10):
    if i % 2 == 0:
        even_data.append(i)
print(even_data)
```

```python
data = [n for n in range(30) if n % 2 == 0 ]
print(data)

# n = output list
```

```python
# syntax 1
# [expression for var in sequence]

# each time it read one value from the sequence. it will apply the expression
# the result of this expression is stored inside in list

list1 = [1, 2, 3, 4, 5]
list2 = [n*n for n in list1]

print(list1)
print(list2)

```

```python
# syntax 2 

# [expression for var in sequence if condition]
# if condiiton is used for filter

list1 = [1, 2, 3, 4, 5]
list2 = [n for n in list1 if n % 2 == 0]
print(list2)
```

```python
# syntax 3

# nested list

# [Expression for var in iterable for var in iterable]

# var in iterable   --- outer forloop
# var in iterable   --- inner forloop

# First python evalutes outer forloop than inner forloop
# the result of the expression shown in the list

list1 = [[1, 2], [3, 4]]
list2 = [item for row in list1 for item in row]
print(list2)
```

```python

# example 2 for nested list in list comprension

list1 = ['a', 'b', 'c', 'd', 'e']
list2 = [1, 2, 3, 4, 5]

list3 = [(val1, val2) for val1 in list1 for val2 in list2]
print(list3)
```
