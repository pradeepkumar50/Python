```python
# Arithmematic operators

x = 5
y = 10

add_result = x + y
sub_result = x - y
mul_result = x * y
div_result = x / y
floor_div_result = x // y
mod_result = x % y
exponent_result = x ** y

print(add_result, sub_result, mul_result, div_result, floor_div_result, mod_result, exponent_result)
```

```python
# comparision ooperators

first_name = "pradeep kumar"
last_name = "kairamkonda"

result = first_name == last_name
print(result)

# Not equal to 
result2 = first_name != last_name
print(result2)

deep_age = 25
nikith_age = 25

result3 = deep_age > nikith_age
print(result3)

result4 = deep_age < nikith_age
print(result4)

result5 = deep_age >= nikith_age
print(result5)

result6 = deep_age <= nikith_age
print(result6)

```

```python
# logical operators
# and --- It returns true if both statements are true
# or ---- It returns true if one of the statement is true
# not --- It reverse the result, returns false if the result is true

age = 5
print(age >= 5 and age == 5)

# or
print(age >=5 or age < 5)

# not
print(not(age >=5 or age < 5))

```
