```python
# key should be unique

data1 = {
    "name": "pradeep kumar",           # the name print only once
    "id": 495,
    "college": "kmit",
    "name": "mani"
}

print(data1)
```

```python
# Accessing dictinory elements

data1 = {
    "name": "pradeep kumar",           
    "id": 495,
    "college": "kmit",
    "grade": "A"
}

print(data1["name"])
print(data1["grade"])

# other way using get method

print(data1.get("name"))
print(data1.get("grade"))
print(data1.get("school"))
print(data1.get("school", "Not Avaliable"))
```

```python
# Modifying the dicitonary elements
# dictionary are mutable, so you can add, updata or delete elements but
# key should be uniqque

data2 = {
    "name": "pradeep kumar",           
    "id": 495,
    "college": "kmit",
    "grade": "A"
}

# update value for the key

data2["grade"] = "C"
print(data2)

# Adding new key and value

data2["address"] = "Illinois"
print(data2)

# Accessing only keys

result = data2.keys()
print(result)

# Get the values

result2 = data2.values()
print(result2)

# Get the items

result3 = data2.items()
print(result3)

# check "name" is present in the dict

if "name" in data2:
    print("yes. name is there in the data2 dictionary")

# update mehtod

result4 = data2.update({"year": 2025})
print(data2)


```

```python
data3 = {
    "name": "pradeep kumar",           
    "id": 495,
    "college": "kmit",
    "grade": "A"
}

# shallow copy

shal_copy = data3

print(data3)
print(shal_copy)

data3["name"] = "arun"

print(data3)
print(shal_copy)

 # shallow copy  provides the different memory locations, so we change one of the variable it should not impact to other one
shal_copy1 = data3.copy()     

print(data3)
print(shal_copy1)


data3["name"] = "John"

print(data3)
print(shal_copy1)
```

```python
# Iterating over ditctories
# you can use loops to iterate dictionries
data4 = {
    "name": "pradeep kumar",           
    "id": 495,
    "college": "kmit",
    "grade": "A"
}

# iterationg over keys

for keys in data4.keys():
    print(keys)

for values in data4.values():
    print(values)

for key, value in data4.items():
    print(key, value)
```

```python
# nested dictronies
warehouse = {
   "payless": {
       "name": "Ritche",
       "id": 495
   },
    "lucky": {
        "name": "john",
        "id": 496
    }
}

print(warehouse)

# Accessing the nested dictinories

print(warehouse["payless"]["id"])

# looping nested dict

for x, obj in warehouse.items():
    print(x)
for y in obj:
    print(y + ':', obj[y])


```

```python
# Dict comphrehension

numbers = {x: x ** 2 for x in range(10)}
print(numbers)

# condition

numbers_even = {x: x ** 2 for x in range(10) if x % 2 == 0}
print(numbers_even)

```
