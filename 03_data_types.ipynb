```python
# Data types

# Data types are a classification of data which tells the compiler or interpreter
# how the programmer intends to use the data.
 
# Importance of data types in programming


# proper use of data types can prevent errors and bugs in the program.
# They help in performing correct operations on data

# Basic data types

# Integers
# Floating point numbers
# strings
# Boolens

# Advanced Data Types

# Lists
# Tuples
# Sets
# Dictionaries

```

```python
# Basic data types
age = 29
print(type(age))

height = 5.7
print(type(height))

name = "pradeep kumar kairamkonda"
print(type(name))

is_student = True
print(type(is_student))

x = 1j
print(type(x))

y = range(6)
print(type(y))

z = None
print(type(z))
```

```python
# Advanced data types

data = ["pradeep", "kumar", "kairamkonda"]
print(type(data))

fruits = ("apple", "banana", "pineapple")
print(type(fruits))

animals = {"donkey", "monkey", "elephant"}
print(type(animals))

student_data = {"name" : "pradeep", "age" : 25}
print(type(student_data))

x = frozenset({"apple", "banana", "cherry"})
print(type(x))
```
